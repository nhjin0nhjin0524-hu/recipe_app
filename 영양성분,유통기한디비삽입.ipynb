```python
import mysql.connector

# ==========================================
# 1. DB 설정 (본인 환경에 맞게 수정)
# ==========================================
DB_CONFIG = {
    "host": "localhost",
    "user": "root",
    "password": "root",
    "database": "cooking_db",
    "charset": "utf8mb4"
}

# ==========================================
# 2. 통합 데이터 (분석 완료된 Seed Data)
# ==========================================
# 형식: {이름, 유통기한(일), 칼로리, 탄수, 단백, 지방}
MY_DATA = [
    {'name': '돼지고기', 'days': 3, 'kcal': 303.0, 'carbs': 0.0, 'protein': 30.5, 'fat': 19.5},
    {'name': '소고기', 'days': 3, 'kcal': 335.0, 'carbs': 0.0, 'protein': 30.1, 'fat': 22.8},
    {'name': '닭고기', 'days': 2, 'kcal': 165.0, 'carbs': 0.0, 'protein': 31.0, 'fat': 3.6},
    {'name': '햄', 'days': 7, 'kcal': 197.0, 'carbs': 0.28, 'protein': 15.43, 'fat': 2.62},
    {'name': '베이컨', 'days': 7, 'kcal': 266.0, 'carbs': 0.14, 'protein': 16.58, 'fat': 2.17},
    {'name': '소시지', 'days': 3, 'kcal': 276.0, 'carbs': 3.12, 'protein': 12.18, 'fat': 23.34},
    {'name': '달걀', 'days': 14, 'kcal': 138.0, 'carbs': 3.6, 'protein': 11.0, 'fat': 8.6},
    {'name': '계란', 'days': 14, 'kcal': 138.0, 'carbs': 3.6, 'protein': 11.0, 'fat': 8.6}, # 동의어 추가
    {'name': '두부', 'days': 3, 'kcal': 97.0, 'carbs': 0.3, 'protein': 9.6, 'fat': 4.6},
    {'name': '순두부', 'days': 3, 'kcal': 50.0, 'carbs': 1.6, 'protein': 6.3, 'fat': 1.8},
    {'name': '어묵', 'days': 3, 'kcal': 146.0, 'carbs': 19.4, 'protein': 10.9, 'fat': 2.8},
    {'name': '게맛살', 'days': 7, 'kcal': 86.0, 'carbs': 12.3, 'protein': 8.5, 'fat': 0.4},
    {'name': '참치통조림', 'days': 1095, 'kcal': 200.0, 'carbs': 0.0, 'protein': 17.5, 'fat': 14.2},
    {'name': '고등어', 'days': 2, 'kcal': 183.0, 'carbs': 0.0, 'protein': 19.0, 'fat': 12.0},
    {'name': '삼치', 'days': 2, 'kcal': 176.0, 'carbs': 0.1, 'protein': 19.3, 'fat': 10.4},
    {'name': '꽁치', 'days': 2, 'kcal': 262.0, 'carbs': 0.0, 'protein': 18.0, 'fat': 20.0},
    {'name': '갈치', 'days': 2, 'kcal': 145.0, 'carbs': 0.0, 'protein': 18.0, 'fat': 7.5},
    {'name': '오징어', 'days': 2, 'kcal': 87.0, 'carbs': 0.0, 'protein': 18.0, 'fat': 1.0},
    {'name': '낙지', 'days': 1, 'kcal': 53.0, 'carbs': 0.0, 'protein': 11.5, 'fat': 0.4},
    {'name': '새우', 'days': 2, 'kcal': 93.0, 'carbs': 0.0, 'protein': 19.0, 'fat': 1.0},
    {'name': '조개', 'days': 1, 'kcal': 60.0, 'carbs': 2.0, 'protein': 9.0, 'fat': 1.0},
    {'name': '멸치', 'days': 7, 'kcal': 288.0, 'carbs': 0.0, 'protein': 55.6, 'fat': 5.4},
    {'name': '김', 'days': 60, 'kcal': 19.0, 'carbs': 3.6, 'protein': 3.0, 'fat': 0.3},
    {'name': '미역', 'days': 1095, 'kcal': 11.0, 'carbs': 3.0, 'protein': 1.0, 'fat': 0.1},
    {'name': '다시마', 'days': 60, 'kcal': 19.0, 'carbs': 6.6, 'protein': 0.8, 'fat': 0.1},
    {'name': '콩나물', 'days': 3, 'kcal': 30.0, 'carbs': 2.3, 'protein': 4.6, 'fat': 1.4},
    {'name': '숙주나물', 'days': 3, 'kcal': 11.0, 'carbs': 1.7, 'protein': 1.8, 'fat': 0.1},
    {'name': '시금치', 'days': 4, 'kcal': 33.0, 'carbs': 4.6, 'protein': 4.6, 'fat': 0.6},
    {'name': '배추', 'days': 14, 'kcal': 13.0, 'carbs': 3.0, 'protein': 1.0, 'fat': 0.1},
    {'name': '양배추', 'days': 21, 'kcal': 33.0, 'carbs': 7.6, 'protein': 1.6, 'fat': 0.1},
    {'name': '상추', 'days': 5, 'kcal': 18.0, 'carbs': 3.5, 'protein': 1.5, 'fat': 0.3},
    {'name': '깻잎', 'days': 4, 'kcal': 41.0, 'carbs': 7.5, 'protein': 4.3, 'fat': 0.7},
    {'name': '무', 'days': 14, 'kcal': 18.0, 'carbs': 3.7, 'protein': 0.9, 'fat': 0.1},
    {'name': '당근', 'days': 14, 'kcal': 34.0, 'carbs': 7.8, 'protein': 0.9, 'fat': 0.1},
    {'name': '양파', 'days': 14, 'kcal': 35.0, 'carbs': 8.5, 'protein': 0.9, 'fat': 0.1},
    {'name': '대파', 'days': 7, 'kcal': 26.0, 'carbs': 6.5, 'protein': 1.5, 'fat': 0.2},
    {'name': '쪽파', 'days': 7, 'kcal': 25.0, 'carbs': 4.5, 'protein': 2.0, 'fat': 0.3},
    {'name': '마늘', 'days': 14, 'kcal': 120.0, 'carbs': 26.3, 'protein': 6.3, 'fat': 0.2},
    {'name': '깐마늘', 'days': 10, 'kcal': 120.0, 'carbs': 26.3, 'protein': 6.3, 'fat': 0.2},
    {'name': '생강', 'days': 14, 'kcal': 53.0, 'carbs': 10.9, 'protein': 1.8, 'fat': 0.6},
    {'name': '고추', 'days': 10, 'kcal': 57.0, 'carbs': 11.2, 'protein': 2.3, 'fat': 0.8},
    {'name': '청양고추', 'days': 10, 'kcal': 57.0, 'carbs': 11.2, 'protein': 2.3, 'fat': 0.8},
    {'name': '오이', 'days': 7, 'kcal': 11.0, 'carbs': 2.3, 'protein': 0.8, 'fat': 0.1},
    {'name': '호박', 'days': 7, 'kcal': 27.0, 'carbs': 6.2, 'protein': 1.4, 'fat': 0.2},
    {'name': '애호박', 'days': 7, 'kcal': 38.0, 'carbs': 7.6, 'protein': 1.3, 'fat': 0.2},
    {'name': '가지', 'days': 7, 'kcal': 17.0, 'carbs': 3.9, 'protein': 0.8, 'fat': 0.1},
    {'name': '토마토', 'days': 7, 'kcal': 19.0, 'carbs': 4.1, 'protein': 0.9, 'fat': 0.2},
    {'name': '방울토마토', 'days': 7, 'kcal': 16.0, 'carbs': 3.2, 'protein': 0.9, 'fat': 0.1},
    {'name': '감자', 'days': 21, 'kcal': 63.0, 'carbs': 14.5, 'protein': 2.2, 'fat': 0.1},
    {'name': '고구마', 'days': 21, 'kcal': 128.0, 'carbs': 30.3, 'protein': 1.4, 'fat': 0.2},
    {'name': '버섯', 'days': 5, 'kcal': 30.0, 'carbs': 6.1, 'protein': 2.7, 'fat': 0.4},
    {'name': '표고버섯', 'days': 5, 'kcal': 30.0, 'carbs': 6.1, 'protein': 2.7, 'fat': 0.4},
    {'name': '느타리버섯', 'days': 5, 'kcal': 22.0, 'carbs': 5.0, 'protein': 2.4, 'fat': 0.3},
    {'name': '팽이버섯', 'days': 5, 'kcal': 32.0, 'carbs': 6.6, 'protein': 2.5, 'fat': 0.3},
    {'name': '새송이버섯', 'days': 5, 'kcal': 24.0, 'carbs': 4.8, 'protein': 2.8, 'fat': 0.3},
    {'name': '쌀', 'days': 365, 'kcal': 360.0, 'carbs': 79.0, 'protein': 7.0, 'fat': 1.0},
    {'name': '찹쌀', 'days': 365, 'kcal': 366.0, 'carbs': 80.0, 'protein': 7.0, 'fat': 0.8},
    {'name': '현미', 'days': 180, 'kcal': 354.0, 'carbs': 74.0, 'protein': 7.5, 'fat': 2.5},
    {'name': '보리', 'days': 365, 'kcal': 334.0, 'carbs': 72.0, 'protein': 10.0, 'fat': 1.8},
    {'name': '밀가루', 'days': 365, 'kcal': 364.0, 'carbs': 76.0, 'protein': 10.0, 'fat': 1.0},
    {'name': '부침가루', 'days': 365, 'kcal': 350.0, 'carbs': 78.0, 'protein': 8.0, 'fat': 1.2},
    {'name': '튀김가루', 'days': 365, 'kcal': 355.0, 'carbs': 77.0, 'protein': 8.0, 'fat': 1.5},
    {'name': '전분', 'days': 365, 'kcal': 350.0, 'carbs': 86.0, 'protein': 0.3, 'fat': 0.1},
    {'name': '당면', 'days': 730, 'kcal': 350.0, 'carbs': 87.0, 'protein': 0.1, 'fat': 0.1},
    {'name': '소면', 'days': 730, 'kcal': 365.0, 'carbs': 76.0, 'protein': 11.0, 'fat': 1.0},
    {'name': '라면', 'days': 180, 'kcal': 450.0, 'carbs': 65.0, 'protein': 9.0, 'fat': 16.0},
    {'name': '파스타면', 'days': 730, 'kcal': 365.0, 'carbs': 74.0, 'protein': 12.0, 'fat': 1.5},
    {'name': '고추장', 'days': 365, 'kcal': 210.0, 'carbs': 45.0, 'protein': 4.0, 'fat': 1.0},
    {'name': '된장', 'days': 365, 'kcal': 128.0, 'carbs': 16.0, 'protein': 12.0, 'fat': 4.0},
    {'name': '쌈장', 'days': 365, 'kcal': 200.0, 'carbs': 35.0, 'protein': 8.0, 'fat': 3.0},
    {'name': '간장', 'days': 500, 'kcal': 53.0, 'carbs': 8.0, 'protein': 6.0, 'fat': 0.0},
    {'name': '소금', 'days': 1800, 'kcal': 0.0, 'carbs': 0.0, 'protein': 0.0, 'fat': 0.0},
    {'name': '설탕', 'days': 1800, 'kcal': 400.0, 'carbs': 100.0, 'protein': 0.0, 'fat': 0.0},
    {'name': '식초', 'days': 1000, 'kcal': 20.0, 'carbs': 5.0, 'protein': 0.0, 'fat': 0.0},
    {'name': '참기름', 'days': 365, 'kcal': 884.0, 'carbs': 0.0, 'protein': 0.0, 'fat': 100.0},
    {'name': '들기름', 'days': 180, 'kcal': 884.0, 'carbs': 0.0, 'protein': 0.0, 'fat': 100.0},
    {'name': '식용유', 'days': 365, 'kcal': 884.0, 'carbs': 0.0, 'protein': 0.0, 'fat': 100.0},
    {'name': '올리브유', 'days': 365, 'kcal': 884.0, 'carbs': 0.0, 'protein': 0.0, 'fat': 100.0},
    {'name': '버터', 'days': 180, 'kcal': 717.0, 'carbs': 0.1, 'protein': 0.9, 'fat': 81.0},
    {'name': '마요네즈', 'days': 240, 'kcal': 700.0, 'carbs': 1.0, 'protein': 1.0, 'fat': 75.0},
    {'name': '케찹', 'days': 300, 'kcal': 120.0, 'carbs': 28.0, 'protein': 1.0, 'fat': 0.0},
    {'name': '고춧가루', 'days': 365, 'kcal': 250.0, 'carbs': 45.0, 'protein': 13.0, 'fat': 10.0},
    {'name': '후추', 'days': 1095, 'kcal': 250.0, 'carbs': 64.0, 'protein': 10.0, 'fat': 3.0},
    {'name': '깨', 'days': 730, 'kcal': 570.0, 'carbs': 23.0, 'protein': 18.0, 'fat': 50.0},
    {'name': '김치', 'days': 60, 'kcal': 18.0, 'carbs': 3.0, 'protein': 2.0, 'fat': 0.0},
    {'name': '단무지', 'days': 30, 'kcal': 15.0, 'carbs': 3.5, 'protein': 0.2, 'fat': 0.1},
    {'name': '우유', 'days': 10, 'kcal': 60.0, 'carbs': 5.0, 'protein': 3.0, 'fat': 3.5},
    {'name': '치즈', 'days': 180, 'kcal': 400.0, 'carbs': 1.0, 'protein': 25.0, 'fat': 33.0},
    {'name': '요구르트', 'days': 14, 'kcal': 60.0, 'carbs': 4.7, 'protein': 3.5, 'fat': 3.3},
    {'name': '빵', 'days': 3, 'kcal': 260.0, 'carbs': 49.0, 'protein': 9.0, 'fat': 3.0},
    {'name': '떡', 'days': 1, 'kcal': 230.0, 'carbs': 52.0, 'protein': 4.0, 'fat': 0.5},
    {'name': '파프리카', 'days': 7, 'kcal': 20.0, 'carbs': 4.7, 'protein': 1.0, 'fat': 0.1},
    {'name': '브로콜리', 'days': 5, 'kcal': 28.0, 'carbs': 5.0, 'protein': 3.0, 'fat': 0.3},
    {'name': '사과', 'days': 21, 'kcal': 57.0, 'carbs': 14.0, 'protein': 0.2, 'fat': 0.1},
    {'name': '배', 'days': 14, 'kcal': 51.0, 'carbs': 11.0, 'protein': 0.5, 'fat': 0.1},
    {'name': '귤', 'days': 7, 'kcal': 39.0, 'carbs': 10.0, 'protein': 0.7, 'fat': 0.1},
    {'name': '포도', 'days': 5, 'kcal': 60.0, 'carbs': 15.0, 'protein': 0.7, 'fat': 0.2},
    {'name': '즉석밥', 'days': 180, 'kcal': 145.0, 'carbs': 32.0, 'protein': 2.5, 'fat': 0.3}
]

def save_hardcoded_data():
    conn = None
    try:
        conn = mysql.connector.connect(**DB_CONFIG)
        cursor = conn.cursor(dictionary=True)
        print("✅ DB 연결 성공! 데이터 주입을 시작합니다...")

        success_count = 0
        
        for item in MY_DATA:
            target_name = item['name']
            
            # 1. DB에서 해당 이름이 포함된 재료 ID 찾기 (스마트 매칭)
            # 예: "소고기" 검색 -> "소고기(국산)", "소고기(수입)" 모두 찾아냄
            # 정확도를 위해 2글자 이상일 때만 LIKE 검색, 1글자는 정확 일치 권장
            if len(target_name) > 1:
                sql_find = "SELECT id, name FROM ingredients WHERE name LIKE %s"
                cursor.execute(sql_find, (f"%{target_name}%",))
            else:
                sql_find = "SELECT id, name FROM ingredients WHERE name = %s"
                cursor.execute(sql_find, (target_name,))
                
            matches = cursor.fetchall()
            
            if not matches:
                # DB에 없는 재료(예: 뱅어포, 하몽 등)는 조용히 넘어갑니다.
                continue
                
            for match in matches:
                ing_id = match['id']
                
                # 2. 유통기한 업데이트 (0보다 클 때만)
                if item['days'] > 0:
                    cursor.execute("UPDATE ingredients SET shelf_life_days = %s WHERE id = %s", 
                                   (item['days'], ing_id))
                
                # 3. 영양성분 저장 (INSERT / UPDATE)
                # 영양성분 값이 하나라도 있으면 저장
                if item['kcal'] > 0 or item['protein'] > 0:
                    sql_nut = """
                        INSERT INTO ingredient_nutrition 
                        (ingredient_id, basis, calories, carbs, protein, fat)
                        VALUES (%s, 'per_100g', %s, %s, %s, %s)
                        ON DUPLICATE KEY UPDATE
                            calories = VALUES(calories),
                            carbs = VALUES(carbs),
                            protein = VALUES(protein),
                            fat = VALUES(fat)
                    """
                    cursor.execute(sql_nut, (
                        ing_id, 
                        item.get('kcal', 0), 
                        item.get('carbs', 0), 
                        item.get('protein', 0), 
                        item.get('fat', 0)
                    ))
                
                success_count += 1

        conn.commit()
        print(f"\n🎉 작업 완료! 총 {success_count}건의 재료 정보가 업데이트되었습니다.")

    except Exception as e:
        print(f"❌ 에러 발생: {e}")
    finally:
        if conn: conn.close()

if __name__ == "__main__":
    save_hardcoded_data()
```
